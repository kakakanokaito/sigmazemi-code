## ④ やってみよう：好きな文を入れて、何トークンになるか数える

```python
import tiktoken
enc = tiktoken.get_encoding("o200k_base")
text = "ずんだもんはずんだ餅が大好きなのだ"   # ← ここを好きな文に変える
ids = enc.encode(text)
print(len(text), "文字 →", len(ids), "トークン")
print(" | ".join(enc.decode_single_token_bytes(i).decode("utf-8", errors="replace") for i in ids))
print(ids)
```
